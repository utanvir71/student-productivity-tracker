# STUDENT PRODUCTIVITY TRANCKER

```python
tasks = []
subjects = set()
PRIORITIES = ("low", "medium", "high")
```

```python
def get_hours():
    while True:
        hours_text = input("Estimate study time in hours (e.g. 1.5) (q to exit): ")
        if hours_text == 'q':
                return 0
        if hours_text.replace(".", "", 1).isdigit():
            hours = float(hours_text)
            if hours>0:
                return hours
        print("Invalid number. Enter a postive number")
```

```python
def add_task():
    while True:
        title = input("Enter task name (q to exit): ").strip()
        if title == 'q':
            return
        if title == "":
            print("Task name cannot be emptly.")
            continue
        break
    while True:
        subject = input("Enter subject (q to exit): ").strip().title()
        if subject == 'q':
                    return
        if subject == "":
            print("subjects cannot be emptty.")
            continue
        break
    while True:
        priority = input("Enter priority (low/medium/high) (q to exit): ").strip().lower()
        if priority == 'q':
                    return
        if priority not in PRIORITIES:
            print("invalid priority. ")
            continue
        break
    hours = get_hours()
    if hours == 0: 
          return
    task = {
        "title": title,
        "subjects" : subject,
        "priority" : priority.title(),
        "hours": hours,
        "completed": False
    }
    tasks.append(task)
    subjects.add(subject)
    print("Task added successfully!")
```

```python
def view_tasks():
    if len(tasks) == 0:
        print("No tasks avaialbe.")
        return
    print("\n--- All Tasks ---")

    for number, task in enumerate(tasks, start = 1):
        if task["completed"]:
            status = "Completed"
        else:
            status = "Pending"
        print(
            f"{number}. {task['title']} | "
            f"Subject: {task['subject']} | "
            f"Priority: {task['priority']} | "
            f"Hours: {task['hours']:.1f} | "
            f"Status: {status}"
        )
```

```python
def mark_task_completed():
    if len(tasks) == 0:
        print("No tasks available.")
        return
    view_tasks()
    while True:
        number = input("Enter task number to complete (q to exit): ")
        if number == 'q':
            break
        if not number.isdigit():
            print("Enter a valid positive whole number.")
            continue
        index = int(number) -1
        if index < 0 or index >= len(tasks):
            print("Task number does not exist")
            continue
        elif tasks[index]["completed"]:
            print("Task already marked as completed.")
            continue
        else: 
            tasks[index]["completed"] = True
            print("task marked as completed!")
            
```

```python
def search_task():
    keyword = input("Enter search keyword: ").strip().lower()
    if keyword == "":
        print("Search keyward cannot be empty.")
    found = False

    for task in tasks:
        if (keyword in task["title"].lower() or keyword in task["subject"].lower()):
            print(f"{task['title']} |"
                  f"{task['subject']}"
                  f"{task['priority']}")
            found = True
        if not found:
            print("No matching tasks found.")
```

```python
def show_summary():
    total = len(tasks)
    completed = 0
    total_hours = 0.0

    for task in tasks:
        total_hours += task["hours"]
        if task["completed"]:
            completed += 1

    pending = total - completed

    print("\n--- Producitvity Summury ----")
    print("Total tasks: ", total)
    print("Pending tasks: ", pending)
    print(f"Estimated study hours: {total_hours:.1f}")

    if subjects:
        print("subjects", ", ".join(sorted(subjects)))
    else:
        print("subjects: None")
```

```python
def main():
    while True:
        print("\n==== STUDENT PRODUCTIVUTY TRACKER ====")
        print("1. Add a task")
        print("2. View all task")
        print("3. Mark a task as completed")
        print("4. Search tasks")
        print("5. View productivity summary")
        print("6. Exit")

        choice = input("Enter you choice: ").strip()

        if choice == "1":
            add_task()
        elif choice == '2':
            view_tasks()
        elif choice == '3':
            mark_task_completed()
        elif choice == '4':
            search_task()
        elif choice == "5":
            show_summary()
        elif choice == "6":
            print("Goodbye! good luck with your studies.")
            break
        else:
            print("Invalid choice. please select 1-6.")
```

```python
main()
```
